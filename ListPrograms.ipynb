```python
numbers=[10,20,30,40,50]
print("List:",numbers)
print("First element:",numbers[0])
print("Second element:",numbers[1])
print("Third element:",numbers[2])
print("Fourth element:",numbers[3])
print("Fifth element:",numbers[4])
```

```python
print("Original List:",numbers)
numbers.append(60)
print("After Appending:",numbers)
```

```python
print("Original List:",numbers)
numbers.reverse()
print("Reversed List:",numbers)
```

```python
#count the occurences
element=int(input("Enter the element to search:"))
count=numbers.count(element)
print("Number of occurences",count)
```

```python
#append list 1 to list 2
list1=(10,20,30)
list2=(40,50,60)
list2=list2+list1
print(list1)
print(list2)
```

```python
#insert element at specified position
print("Original List:",numbers)
numbers.insert(1,15)
print("After Insertion:",numbers)
```

```python
index=int(input("Enter the index to remove: "))
numbers.pop(index)
print(numbers)
```

```python
for i in range(20):
    num=int(input("Enter value:"))
    numbers.append(num)
print("\nList:",numbers)
#a)Count similar elements and print their index values
print("\nSimilar elements and their index values:")
for i in range(20):
    for j in range(20):
        if numbers[j]==numbers[i]:
            print(j,end=" ")
    print()
#b)Count even and odd values
even_count=0
odd_count=0

for num in numbers:
    if num%2==0:
        even_count+=1
    else:
        odd_count+=1
print("\nEven values:",even_count)
print("Odd values:",odd_count)
#c) Count positive and negative values
positive_count=0
negative_count=0
for num in numbers:
    if num>0:
        positive_count+=1
    else:
        negative_count+=1
print("\nPositive values:",positive_count)
print("Negative values:",negative_count)
    
```

```python
#wap to iterate over a dictionary usimg for loop and print keys alone, values alone and both tgt
a={101:"s",102:"e",103:"w"}
print("keys:")
for key in a.keys():
    print(key)
print("\nvalues")
for value in a.values():
    print(value)
print("\nkeys and values")
for key,value in a.items():
    print(key,value)
```

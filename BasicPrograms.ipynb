```python
a=float(input("Enter the first number"))
b=float(input("Enter the second number"))
operation=(input("Enter the operation symbol"))
if operation=="+":
    print("Answer: ",a+b)
elif operation=="-":
    print("Answer: ",a-b)
elif operation=="*":
    print("Answer: ",a*b)
elif operation=="/":
    if b==0:
        print("Not allowed")
        quit()
    else:
        print("Answer: ",(a/b))
elif operation=="%":
    print("Answer: ",(a%b))
else:
    print("Not a valid operation")
```
